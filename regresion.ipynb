```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
# PREDICCIÓN DE PRECIOS DE CASAS

# 1. Definir datos de ejemplo
data = {
    'surface': [1100, 1200 ,1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000],
    'price': [120000, 140000, 160000, 180000, 200000, 220000, 240000, 260000, 280000, 300000]
}
df = pd.DataFrame(data)

# 2. Identificación de variables independientes (características) y dependientes
X = df[['surface']]  # Variable independiente
y = df['price']      # Variable dependiente

# 3. División de los datos en conjuntos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# 4. Creación del modelo de regresión lineal
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# 5. Predicción de precios en el conjunto de prueba
y_pred = model.predict(X_test)

# 6. Evaluación del modelo
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f'Mean Squared Error: {mse}')
print(f'R² Score: {r2}')
```

```python
# 7. Visualización de los resultados
plt.scatter(X, y, color='blue', label='Datos reales')
plt.plot(X_test, y_pred, color='red', linewidth=2, label='Predicciones')
plt.xlabel('Superficie (m²)')
plt.ylabel('Precio ($)')
plt.title('Regresión Lineal')
plt.legend()
plt.show()
```
