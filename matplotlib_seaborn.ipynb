```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'surface': [1100, 1200 ,1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000],
    'price': [120000, 140000, 160000, 180000, 200000, 220000, 240000, 260000, 280000, 300000]
}
df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='surface', y='price', data=df)
plt.title('Regresión  lineal: Surface vs Price')
plt.xlabel('Surface')
plt.ylabel('Price')
```

```python
sns.regplot(x='surface', y='price', data=df, ci=None, scatter_kws={'color': 'blue'}, line_kws={'color': 'red'})
plt.show()
```
