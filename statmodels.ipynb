```python
import pandas as pd
import statmodels.api as sm
```

```python
# PREDICCIÓN DE PRECIOS DE CASAS

# 1. Definir datos de ejemplo
data = {
    'surface': [1100, 1200 ,1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000],
    'price': [120000, 140000, 160000, 180000, 200000, 220000, 240000, 260000, 280000, 300000]
}
df = pd.DataFrame(data)

# 2. Identificación de variables independientes (características) y dependientes
X = df[['surface']]  # Variable independiente
y = df['price']      # Variable dependiente

# Añadir una constante para el modelo de regresión
X = sm.add_constant(X)

# Crear modelo de regresión lineal
model = sm.OLS(y, X).fit()

# Imprimir resumen del modelo
print(model.summary())

```
