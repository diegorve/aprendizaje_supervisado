```python
import numpy as np
import scipy.stats as sp
```

```python
surface = np.array([1100, 1200 ,1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000])
price = np.array([120000, 140000, 160000, 180000, 200000, 220000, 240000, 260000, 280000, 300000])

pendiente, interseccion, r_value, p_value, std_err = sp.linregress(surface, price)

print(f"Pendiente: {pendiente}")
print(f"Intersección: {interseccion}")
print(f"Coeficinete de correlación: {r_value}")
print(f"Valor p: {p_value}")
print(f"Error estándar: {std_err}")
```
